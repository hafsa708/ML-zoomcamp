```python
# 1. Median before filling
med_before = df['horsepower'].median()

# 2. Most frequent value (mode)
mode_val = df['horsepower'].mode()[0]

# 3. Fill missing values with the mode and recalculate median
med_after = df['horsepower'].fillna(mode_val).median()

print(f"Before: {med_before}")
print(f"After : {med_after}")
```

```python
import numpy as np

# 1. Select cars from Asia
asia_df = df[df['origin'] == 'Asia']

# 2. Select vehicle_weight and model_year, first 7 values
X = asia_df[['vehicle_weight', 'model_year']].head(7).to_numpy()

# 3. Matrix-matrix multiplication: X.T * X
XTX = X.T.dot(X)

# 4. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 5. Create y array
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 6. Compute w = (XTX)^(-1) * X.T * y
w = XTX_inv.dot(X.T).dot(y)

# 7. Sum of all elements of w
print("Sum of weights:", w.sum())
```
